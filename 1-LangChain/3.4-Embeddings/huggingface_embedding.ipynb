```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['HF_TOKEN'] = os.getenv("HF_TOKEN")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(
    model_name = 'all-MiniLM-L6-v2'
)

embedding
```

```python
text = "Hello !! My name is Sandeep Hidellarachchi, I am from Sri Lanka and currently live in Leipzig, Germany. This is an atempt to embed text using Huggingface embeddings"

query_results = embedding.embed_query(text)
query_results
```

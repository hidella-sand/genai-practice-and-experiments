```python
# Ollama is just like openAI but this is Open souce therefore we dont need to pay for the embeddings
# Lets import this first
from langchain_community.embeddings import OllamaEmbeddings

# now initialzie the embeddings
embedding = OllamaEmbeddings(
    model='gemma:2b' # for this to work we need to download this gemma 2:b modeel to our comptuer first to do thatw can just run this in cmb "ollama run gemma:2b"
)

embedding
```

```python
# Lets embed a sample sentence

text = "Hello !! My name is Sandeep Hidellarachchi, I am from Sri Lanka and currently live in Leipzig, Germany. This is an atempt to embed text using ollama Embeddings with embed_document"

# Lets embed. embed documents meanas we can give more than one documents ina list format to embed

r1 = embedding.embed_documents(text)
r1
```

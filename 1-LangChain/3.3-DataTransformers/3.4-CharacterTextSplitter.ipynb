# Lets split the documents using Character Text Splitter

```python
from langchain_community.document_loaders import TextLoader

text_loader = TextLoader('speech.txt')

docs = text_loader.load()
docs
```

```python
from langchain_text_splitters import CharacterTextSplitter

c_text_splitter = CharacterTextSplitter(
    separator = "\n\n",
    chunk_size = 100,
    chunk_overlap = 50
)

final_docs = c_text_splitter.split_documents(documents=docs)

# in here we get a message like "Created a chunk of size 470, which is longer than the specified 100" because 
# the documnet does not have a seperators we used in the spliiter so it went to 470 charcatsr and then found a seperator
final_docs
```

```python
from langchain_text_splitters import CharacterTextSplitter

a_text_splitter = CharacterTextSplitter(
    separator = "a",
    chunk_size = 100,
    chunk_overlap = 50
)

final_docs_A = a_text_splitter.split_documents(documents=docs)

# in here as we can see i gave the seperator as "a" therefore this gonna split from every a it found withing 100 chunk
final_docs_A
```

# Split the World of warcraft book

```python
# Read the WOW book
from langchain_community.document_loaders import PyPDFLoader

wow_loader = PyPDFLoader('WOW-Arthas-Rise-Of-he-Lich-King.pdf')

wow_docs = wow_loader.load()
wow_docs
```

```python
len(wow_docs)
```

```python
from langchain_text_splitters import CharacterTextSplitter, RecursiveCharacterTextSplitter

wow_c_spliiter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size = 1000,
    chunk_overlap = 100
)

wow_final_c_docs = wow_c_spliiter.split_documents(wow_docs)
wow_final_c_docs
```

```python
wow_r_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap = 100
)

wow_final_r_docs = wow_r_splitter.split_documents(wow_docs)
wow_final_r_docs
```

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')

# this is used for our langsmith tracking
os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')
os.environ['LANGCHAIN_PROJECT'] = os.getenv('LANGCHAIN_PROJECT')

os.environ['LANGCHAIN_TRACING_V2'] = 'true'
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model='gpt-4o'
)
print(llm)
```

```python
# Get input and answers
results = llm.invoke('What is generative AI')
print(results)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ('system','You are an expert AI Engineer. Provide me answers based on the questions but in a the form of shakespeare'),
        ('user','{input}')
    ]
)

prompt
```

```python
chain = prompt|llm


response = chain.invoke({'input':'What is LLM hallucination'})
print(response)
```

```python
# String output parse
from langchain_core.output_parsers import StrOutputParser

output_parse = StrOutputParser()
chain = prompt|llm|output_parse


response = chain.invoke({'input':'What is bias variance trade off'})
print(response)
```

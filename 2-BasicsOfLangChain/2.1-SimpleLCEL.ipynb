```python
import os
from dotenv import load_dotenv
load_dotenv()

# config my api keys
import openai
openai.api_key = os.getenv('OPENAI_API_KEY')
groq_api_key = os.getenv('GROQ_API_KEY')
```

```python
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI

model = ChatGroq(model='llama-3.1-8b-instant', groq_api_key=groq_api_key)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage


messages1 = [
    SystemMessage(content="Translate the given sentece from English to German and Sinhala"),
    HumanMessage(content="Hello My Name is Sandeep Hidellarachchi")
]

result_1 = model.invoke(messages1)
result_1
```

```python
messages2 = [
    SystemMessage(content="Convert the given sentence to ascii code"),
    HumanMessage(content="Hello My Name is Sandeep Hidellarachchi")
]

result_2 = model.invoke(messages2)
result_2
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
parser.invoke(result_2)
```

```python
parser.invoke(result_1)
```

# Using LCEL we can chain the component

```python
messages3 = [
    SystemMessage(content="Rewrite the given sentence in asscending order depend on the letters of each words"),
    HumanMessage(content="Hello My Name is Sandeep Hidellarachchi")
]
```

```python
chain = model|parser
chain.invoke(messages3)
```

# Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
generic_template = 'Translate the following into {language}'

prompt = ChatPromptTemplate.from_messages(
    [('system',generic_template),
     ('user','{text}')
     ]
)
```

```python
result = prompt.invoke({"language":'Hebrew', "text":'Hello My name is sandeep Hidellarachchi and I love Artificial intelligence'})
```

```python
result.to_messages()
```

```python
chain = prompt|model|parser

chain.invoke({"language":'Hebrew', "text":'Hello My name is sandeep Hidellarachchi and I love Artificial intelligence'})
```

```python
chain = prompt|model|parser

chain.invoke({"language":'Sinhala', "text":'That elephant is the biggest elephant i have ever seen'})
```

```python
chain.invoke({"language":'German', "text":'Nissan G-TR is my favourit car and Land rover Defender is my favourit SUV'})
```

```python
chain.invoke({'language':'Latin', 'text':'Today is a great day'})
```
